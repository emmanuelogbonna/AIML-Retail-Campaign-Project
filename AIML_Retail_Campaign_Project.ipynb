```python
# ============================================================
# CUSTOMER FEATURE ENGINEERING
# ============================================================

import pandas as pd

# Make sure InvoiceDate is datetime
transactions["InvoiceDate"] = pd.to_datetime(
    transactions["InvoiceDate"],
    errors="coerce"
)

# Calculate transaction-level sales
transactions["Sales"] = (
    transactions["Quantity"] *
    transactions["UnitPrice"]
)


# ============================================================
# FEATURE 1: TOTAL SALES
# ============================================================

customer_sales = (
    transactions
    .groupby("CustomerID")["Sales"]
    .sum()
    .reset_index(name="total_sales")
)


# ============================================================
# FEATURE 2: UNIQUE PRODUCTS
# ============================================================

customer_products = (
    transactions
    .groupby("CustomerID")["StockCode"]
    .nunique()
    .reset_index(name="unique_products")
)


# ============================================================
# FEATURE 3: NUMBER OF INVOICES
# ============================================================

customer_invoices = (
    transactions
    .groupby("CustomerID")["InvoiceNo"]
    .nunique()
    .reset_index(name="number_of_invoices")
)


# ============================================================
# ADDITIONAL FEATURE 4: PURCHASE DAYS
# ============================================================

# Create date-only column
transactions["PurchaseDate"] = transactions["InvoiceDate"].dt.date

customer_days = (
    transactions
    .groupby("CustomerID")["PurchaseDate"]
    .nunique()
    .reset_index(name="purchase_days")
)


# ============================================================
# ADDITIONAL FEATURE 5: AVERAGE INVOICE VALUE
# ============================================================

invoice_sales = (
    transactions
    .groupby(["CustomerID", "InvoiceNo"])["Sales"]
    .sum()
    .reset_index()
)

average_invoice = (
    invoice_sales
    .groupby("CustomerID")["Sales"]
    .mean()
    .reset_index(name="average_invoice_value")
)


# ============================================================
# ADDITIONAL FEATURE 6: AVERAGE QUANTITY PER INVOICE
# ============================================================

invoice_quantity = (
    transactions
    .groupby(["CustomerID", "InvoiceNo"])["Quantity"]
    .sum()
    .reset_index()
)

average_quantity = (
    invoice_quantity
    .groupby("CustomerID")["Quantity"]
    .mean()
    .reset_index(name="average_quantity_per_invoice")
)


# ============================================================
# ADDITIONAL FEATURE 7: AVERAGE PRODUCT PRICE
# ============================================================

average_price = (
    transactions
    .groupby("CustomerID")["UnitPrice"]
    .mean()
    .reset_index(name="average_product_price")
)


# ============================================================
# ADDITIONAL FEATURE 8: CUSTOMER LIFETIME
# ============================================================

customer_lifetime = (
    transactions
    .groupby("CustomerID")["InvoiceDate"]
    .agg(["min", "max"])
    .reset_index()
)

customer_lifetime["customer_lifetime_days"] = (
    customer_lifetime["max"] -
    customer_lifetime["min"]
).dt.days

customer_lifetime = customer_lifetime[
    ["CustomerID", "customer_lifetime_days"]
]


# ============================================================
# COMBINE ALL FEATURES
# ============================================================

customer_features = (
    customer_sales
    .merge(customer_products, on="CustomerID", how="left")
    .merge(customer_invoices, on="CustomerID", how="left")
    .merge(customer_days, on="CustomerID", how="left")
    .merge(average_invoice, on="CustomerID", how="left")
    .merge(average_quantity, on="CustomerID", how="left")
    .merge(average_price, on="CustomerID", how="left")
    .merge(customer_lifetime, on="CustomerID", how="left")
)

print("Customer feature dataset shape:", customer_features.shape)

display(customer_features.head(10))
```
